```python
"""
Magnetotransport and Topologically Protected Edge States in Graphene Nanoribbons:
A Study of the Hofstadter Butterfly Spectrum

This script computes and visualizes:
  1. Hofstadter Butterfly spectrum for a zigzag graphene nanoribbon (ZGNR)
  2. Relativistic Landau levels for Dirac fermions
  3. Topologically protected edge states (zigzag flat band + chiral modes)
  4. Hall conductance via TKNN / Chern numbers
  5. Magnetotransport: longitudinal (Rxx) and transverse (Rxy) resistance
"""

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.colors import LinearSegmentedColormap
from scipy.linalg import eigvalsh
import warnings
warnings.filterwarnings("ignore")

# ─── Physical constants ───────────────────────────────────────────────────────
t   = 2.7        # nearest-neighbour hopping (eV)
a   = 0.246e-9   # graphene lattice constant (m)
e   = 1.6e-19    # electron charge (C)
hbar = 1.055e-34 # reduced Planck (J·s)
h   = 2 * np.pi * hbar

# ─── 1. HOFSTADTER BUTTERFLY  ─────────────────────────────────────────────────
def build_gnr_hamiltonian(N, phi, kx):
    """
    Build the tight-binding Hamiltonian for an N-site zigzag GNR unit cell
    at crystal momentum kx and flux quanta phi = Φ/Φ₀ per hexagon.

    Uses Peierls substitution:  t_ij → t · exp(i · 2π · phi · m)
    for hopping along y (between rows m).
    """
    H = np.zeros((2 * N, 2 * N), dtype=complex)

    for m in range(N):
        # Intra-dimer coupling (A–B within same row)
        peierls = np.exp(1j * 2 * np.pi * phi * m)
        H[2*m,   2*m+1] = -t * peierls
        H[2*m+1, 2*m  ] = -t * np.conj(peierls)

        # Inter-dimer coupling (B in row m → A in row m+1)
        if m < N - 1:
            peierls2 = np.exp(1j * 2 * np.pi * phi * (m + 0.5))
            H[2*m+1, 2*(m+1)  ] = -t * peierls2
            H[2*(m+1), 2*m+1  ] = -t * np.conj(peierls2)

    # kx-dependent hopping (periodic boundary along x)
    H[0,   2*N-1] += -t * np.exp(-1j * kx)
    H[2*N-1, 0  ] += -t * np.exp( 1j * kx)

    return H


def compute_butterfly(N=20, n_phi=200, n_kx=30):
    """Compute the Hofstadter butterfly spectrum for a ZGNR."""
    phi_vals = np.linspace(0, 1, n_phi)
    kx_vals  = np.linspace(-np.pi, np.pi, n_kx, endpoint=False)

    phi_plot = []
    E_plot   = []

    for phi in phi_vals:
        eigs_phi = []
        for kx in kx_vals:
            H = build_gnr_hamiltonian(N, phi, kx)
            eigs = eigvalsh(H)
            eigs_phi.extend(eigs)
        phi_plot.extend([phi] * len(eigs_phi))
        E_plot.extend(eigs_phi)

    return np.array(phi_plot), np.array(E_plot)


# ─── 2. LANDAU LEVELS (Dirac fermions) ───────────────────────────────────────
def landau_levels(phi_vals, n_max=5):
    """
    Relativistic Landau levels for massless Dirac fermions:
        E_n = sign(n) * v_F * sqrt(2 * e * hbar * B * |n|)
    In tight-binding units:
        E_n ≈ ±t * sqrt(6π |n| φ)
    """
    levels = {}
    for n in range(-n_max, n_max + 1):
        E_n = np.sign(n) * t * np.sqrt(6 * np.pi * np.abs(n) * phi_vals + 1e-12)
        levels[n] = E_n
    return levels


# ─── 3. EDGE STATES (zigzag flat band) ────────────────────────────────────────
def compute_edge_states(N=20, phi=0.1, n_kx=200):
    """
    Compute the band structure at fixed phi; identify edge-localised states
    by their probability weight on boundary sites.
    """
    kx_vals = np.linspace(-np.pi, np.pi, n_kx)
    bands   = np.zeros((n_kx, 2 * N))
    weights = np.zeros((n_kx, 2 * N))  # edge localisation weight

    for i, kx in enumerate(kx_vals):
        H = build_gnr_hamiltonian(N, phi, kx)
        evals, evecs = np.linalg.eigh(H)
        bands[i] = evals
        # Edge weight: probability on first + last 2 sites
        for j in range(2 * N):
            edge_sites = [0, 1, 2*N-2, 2*N-1]
            weights[i, j] = np.sum(np.abs(evecs[edge_sites, j])**2)

    return kx_vals, bands, weights


# ─── 4. CHERN NUMBERS (TKNN) ─────────────────────────────────────────────────
def chern_number_from_tknn(p, q):
    """
    Solve the Diophantine equation  p·s + q·t = 1  for s (Chern number)
    using the TKNN formula. Returns Chern number s for each of q gaps.
    """
    # Extended Euclidean algorithm
    def ext_gcd(a, b):
        if b == 0:
            return a, 1, 0
        g, x, y = ext_gcd(b, a % b)
        return g, y, x - (a // b) * y

    chern_numbers = []
    for gap_idx in range(1, q):
        # For gap r, solve: p·s ≡ r (mod q)
        g, s, _ = ext_gcd(p, q)
        s = (s * gap_idx) % q
        if s > q // 2:
            s -= q
        chern_numbers.append(s)
    return chern_numbers


# ─── 5. MAGNETOTRANSPORT ──────────────────────────────────────────────────────
def magnetotransport(phi_vals, T=1.0, delta=0.01):
    """
    Compute longitudinal (σxx) and Hall (σxy) conductances using
    a simplified Kubo formula model for a ZGNR.

    Hall conductance σxy = C · e²/h where C is integer (Chern number).
    σxx peaks at Landau level energies (broadened by disorder Γ=delta·t).
    """
    E_F = 0.0     # Fermi energy at charge neutrality point
    sigma_xy = []
    sigma_xx = []
    R_xx     = []
    R_xy     = []

    for phi in phi_vals:
        # Number of filled LLs below E_F
        # For Dirac LLs: E_n ≈ ±t*sqrt(6π|n|φ)
        n_filled = 0
        for n in range(1, 10):
            E_n = t * np.sqrt(6 * np.pi * n * phi)
            if E_n < -E_F:   # negative LL
                n_filled -= 1
            if -E_n < E_F:   # positive LL
                n_filled += 1

        # σxy = ν · e²/h, with ν = 4(n+1/2) for graphene (4-fold degeneracy)
        nu = 4 * n_filled
        sxy = nu   # in units of e²/h

        # σxx: Lorentzian peaks at each LL (broadening Γ)
        sxx = 0.0
        for n in range(-5, 6):
            E_n = np.sign(n) * t * np.sqrt(6 * np.pi * np.abs(n) * phi + 1e-6)
            sxx += delta / ((E_F - E_n)**2 + delta**2)
        sxx *= (e**2 / h)

        sigma_xy.append(sxy)
        sigma_xx.append(sxx)

        # Resistivity tensor inversion: ρ = σ⁻¹
        sxy_SI = sxy * (e**2 / h)
        denom  = sxy_SI**2 + sxx**2 + 1e-30
        R_xx.append(sxx / denom)
        R_xy.append(sxy_SI / denom)

    return (np.array(sigma_xx), np.array(sigma_xy),
            np.array(R_xx),     np.array(R_xy))


# ─── PLOTTING ─────────────────────────────────────────────────────────────────
def make_colormap():
    colors = ['#534AB7', '#EF9F27', '#0F6E56']   # purple → amber → teal
    return LinearSegmentedColormap.from_list('gnr', colors)


def plot_all():
    print("Computing Hofstadter butterfly  (N=20, 200 φ-values) …")
    phi_b, E_b = compute_butterfly(N=20, n_phi=200, n_kx=25)

    print("Computing edge states at φ=0.1 …")
    kx_e, bands_e, weights_e = compute_edge_states(N=20, phi=0.1)

    phi_vals = np.linspace(0.005, 1.0, 400)
    print("Computing Landau levels …")
    ll = landau_levels(phi_vals, n_max=4)

    print("Computing magnetotransport …")
    sxx, sxy, Rxx, Rxy = magnetotransport(phi_vals)

    cmap = make_colormap()

    fig = plt.figure(figsize=(16, 14), facecolor='#0d0d0b')
    fig.suptitle(
        "Magnetotransport & Topologically Protected Edge States\nin Graphene Nanoribbons — Hofstadter Butterfly Spectrum",
        color='white', fontsize=14, fontweight='bold', y=0.98
    )

    gs = gridspec.GridSpec(3, 2, figure=fig,
                           hspace=0.42, wspace=0.32,
                           left=0.08, right=0.96, top=0.93, bottom=0.06)

    ax1 = fig.add_subplot(gs[0, :])   # Butterfly (full width)
    ax2 = fig.add_subplot(gs[1, 0])   # Edge states band structure
    ax3 = fig.add_subplot(gs[1, 1])   # Landau levels
    ax4 = fig.add_subplot(gs[2, 0])   # Hall conductance
    ax5 = fig.add_subplot(gs[2, 1])   # Longitudinal resistance

    dark_bg = '#111110'
    for ax in [ax1, ax2, ax3, ax4, ax5]:
        ax.set_facecolor(dark_bg)
        ax.tick_params(colors='#aaa9a3', labelsize=8)
        for spine in ax.spines.values():
            spine.set_edgecolor('#333330')

    # ── Panel 1: Hofstadter Butterfly ────────────────────────────────────────
    emin, emax = E_b.min(), E_b.max()
    norm_e = (E_b - emin) / (emax - emin)
    colors_b = cmap(norm_e)

    ax1.scatter(phi_b, E_b, c=colors_b, s=0.3, alpha=0.6, rasterized=True)
    ax1.axhline(0, color='#ffffff44', lw=0.8, ls='--')
    ax1.set_xlabel('Magnetic flux per unit cell  φ = Φ/Φ₀', color='#ccc', fontsize=9)
    ax1.set_ylabel('Energy  E  (eV)', color='#ccc', fontsize=9)
    ax1.set_title('Hofstadter Butterfly — ZGNR (N = 20)', color='white', fontsize=10)
    ax1.set_xlim(0, 1)
    ax1.set_ylim(emin * 1.05, emax * 1.05)

    # Annotate rational flux values
    for p, q, label in [(1,3,'1/3'), (1,4,'1/4'), (1,2,'1/2'),
                         (2,3,'2/3'), (3,4,'3/4')]:
        ax1.axvline(p/q, color='#ffffff20', lw=0.6)
        ax1.text(p/q, emax * 0.88, label, color='#ffffff60',
                 ha='center', fontsize=7)

    # ── Panel 2: Edge State Band Structure ───────────────────────────────────
    for j in range(bands_e.shape[1]):
        w = weights_e[:, j]
        ax2.scatter(kx_e / np.pi, bands_e[:, j],
                    c=cmap(w), s=1.2, alpha=0.7, rasterized=True)

    ax2.axhline(0,  color='#ffffff44', lw=0.8, ls='--')
    ax2.axvline(-2/3, color='#ffffff30', lw=0.6)
    ax2.axvline( 2/3, color='#ffffff30', lw=0.6)
    ax2.text(-2/3, bands_e.max()*0.8, 'K', color='#aaa', ha='center', fontsize=7)
    ax2.text( 2/3, bands_e.max()*0.8, "K'", color='#aaa', ha='center', fontsize=7)
    ax2.set_xlabel('Crystal momentum  kₓ / π', color='#ccc', fontsize=9)
    ax2.set_ylabel('Energy  E  (eV)', color='#ccc', fontsize=9)
    ax2.set_title('Band Structure at φ = 0.1\n(colour = edge localisation weight)',
                  color='white', fontsize=9)
    ax2.set_xlim(-1, 1)

    # Colorbar for edge weight
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(0, 1))
    cb = plt.colorbar(sm, ax=ax2, pad=0.02)
    cb.set_label('Edge weight', color='#aaa', fontsize=8)
    cb.ax.yaxis.set_tick_params(color='#aaa', labelsize=7)
    plt.setp(cb.ax.yaxis.get_ticklabels(), color='#aaa')

    # ── Panel 3: Landau Levels ────────────────────────────────────────────────
    ll_colors = {
        0:  '#EF9F27',
        1:  '#5DCAA5',  -1: '#5DCAA5',
        2:  '#7F77DD',  -2: '#7F77DD',
        3:  '#D85A30',  -3: '#D85A30',
        4:  '#378ADD',  -4: '#378ADD',
    }
    for n, E_n in ll.items():
        mask = np.abs(E_n) < 3 * t
        col  = ll_colors.get(n, '#888')
        lbl  = f'n = {n}' if n >= 0 else None
        ax3.plot(phi_vals[mask], E_n[mask], color=col, lw=1.3,
                 alpha=0.9, label=lbl)

    ax3.axhline(0, color='#ffffff55', lw=1.2, ls='--', label='n = 0  (E=0)')
    ax3.set_xlabel('φ = Φ/Φ₀', color='#ccc', fontsize=9)
    ax3.set_ylabel('Landau level energy  E  (eV)', color='#ccc', fontsize=9)
    ax3.set_title('Relativistic Landau Levels\nEₙ = ±v_F √(2eℏB|n|)', color='white', fontsize=9)
    ax3.set_xlim(0, 1)
    ax3.set_ylim(-3*t, 3*t)
    leg = ax3.legend(fontsize=7, loc='upper left', framealpha=0.2,
                     labelcolor='white', facecolor='#111')
    leg.get_frame().set_edgecolor('#333')

    # ── Panel 4: Hall conductance σxy / (e²/h) ───────────────────────────────
    ax4.plot(phi_vals, sxy, color='#7F77DD', lw=1.4, alpha=0.9)
    ax4.axhline(0,  color='#ffffff22', lw=0.6)
    for nu in [-6, -2, 2, 6, 10]:
        ax4.axhline(nu, color='#ffffff15', lw=0.5, ls=':')
        ax4.text(0.98, nu, f'ν={nu}', color='#ffffff44',
                 ha='right', va='center', fontsize=6.5)
    ax4.set_xlabel('φ = Φ/Φ₀', color='#ccc', fontsize=9)
    ax4.set_ylabel('σ_xy  (e²/h)', color='#ccc', fontsize=9)
    ax4.set_title('Quantum Hall Conductance\n(integer plateaus — TKNN Chern #)',
                  color='white', fontsize=9)
    ax4.set_xlim(0, 1)

    # ── Panel 5: Longitudinal resistance Rxx ─────────────────────────────────
    ax5.plot(phi_vals, Rxx * 1e3, color='#F0997B', lw=1.2, alpha=0.9,
             label='R_xx (mΩ)')
    ax5_r = ax5.twinx()
    ax5_r.plot(phi_vals, np.abs(Rxy), color='#5DCAA5', lw=1.1,
               alpha=0.7, ls='--', label='|R_xy| (Ω)')
    ax5_r.tick_params(colors='#aaa9a3', labelsize=8)
    ax5_r.set_ylabel('|R_xy|  (Ω)', color='#5DCAA5', fontsize=8)
    for spine in ax5_r.spines.values():
        spine.set_edgecolor('#333330')

    ax5.set_xlabel('φ = Φ/Φ₀', color='#ccc', fontsize=9)
    ax5.set_ylabel('R_xx  (mΩ)', color='#F0997B', fontsize=9)
    ax5.set_title('Magnetotransport: Shubnikov–de Haas\n(R_xx oscillations, Hall R_xy)',
                  color='white', fontsize=9)
    ax5.set_xlim(0, 1)

    lines1, labs1 = ax5.get_legend_handles_labels()
    lines2, labs2 = ax5_r.get_legend_handles_labels()
    leg2 = ax5.legend(lines1 + lines2, labs1 + labs2, fontsize=7,
                      loc='upper right', framealpha=0.2,
                      labelcolor='white', facecolor='#111')
    leg2.get_frame().set_edgecolor('#333')

    # ── TKNN annotation ───────────────────────────────────────────────────────
    fig.text(0.5, 0.005,
             "TKNN theorem: σ_xy = Σ_n C_n · e²/h   |   "
             "Chern #  C_n from Diophantine eq.  p·s + q·t = 1   |   "
             "Edge states protected by bulk topology",
             ha='center', color='#555550', fontsize=7.5)

    plt.savefig('/mnt/user-data/outputs/hofstadter_gnr.png',
                dpi=180, bbox_inches='tight', facecolor=fig.get_facecolor())
    print("Saved → hofstadter_gnr.png")
    plt.close()


# ─── CHERN NUMBER TABLE ───────────────────────────────────────────────────────
def print_chern_table():
    print("\n── TKNN Chern numbers for rational flux φ = p/q ──")
    print(f"{'p/q':>8}  {'Gap index':>10}  {'Chern #':>9}")
    print("─" * 34)
    for p, q in [(1,3),(1,4),(1,5),(2,5),(1,2),(3,5),(2,3),(3,4)]:
        from math import gcd
        if gcd(p, q) != 1:
            continue
        cherns = chern_number_from_tknn(p, q)
        for idx, C in enumerate(cherns, 1):
            print(f"  {p}/{q}     gap {idx:>3}          C = {C:>+3}")
    print()


# ─── MAIN ─────────────────────────────────────────────────────────────────────
if __name__ == '__main__':
    print_chern_table()
    plot_all()
    print("Done.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------------------------------------------------
# PARAMETERS
# -----------------------------------------------------------------------------
q_max = 60        # resolution
k_points = 80     # momentum sampling

# -----------------------------------------------------------------------------
# HARPER (HOFSTADTER) HAMILTONIAN
# -----------------------------------------------------------------------------
def harper_hamiltonian(k, phi, q):
    """
    Hofstadter Hamiltonian (square lattice)
    Size: q x q
    """
    H = np.zeros((q, q), dtype=complex)

    for m in range(q):
        # onsite modulation (magnetic field effect)
        H[m, m] = 2 * np.cos(2 * np.pi * phi * m + k)

        # nearest-neighbor hopping
        if m < q - 1:
            H[m, m+1] = 1
            H[m+1, m] = 1

    # periodic boundary condition
    H[0, q-1] = 1
    H[q-1, 0] = 1

    return H

# -----------------------------------------------------------------------------
# COMPUTE BUTTERFLY
# -----------------------------------------------------------------------------
phis = []
energies = []

for q in range(1, q_max + 1):
    for p in range(q + 1):

        phi = p / q

        for k in np.linspace(0, 2*np.pi, k_points):

            H = harper_hamiltonian(k, phi, q)
            eigvals = np.linalg.eigvalsh(H)

            phis.extend([phi] * len(eigvals))
            energies.extend(eigvals)

# -----------------------------------------------------------------------------
# PLOT
# -----------------------------------------------------------------------------
plt.figure(figsize=(10, 6))
plt.scatter(phis, energies, s=0.3)

plt.xlabel(r"Magnetic Flux $\phi = p/q$")
plt.ylabel("Energy")
plt.title("Hofstadter Butterfly (Exact Harper Model)", fontweight='bold')

plt.xlim(0, 1)
plt.ylim(-4, 4)

plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()

plt.savefig("figsDissOrder6.png", dpi=300, bbox_inches='tight')
plt.close()
```

```python
import kwant
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# PARAMETERS
# ============================================================

R_outer = 20
R_inner = 10
W_lead = 4

# ============================================================
# LATTICE
# ============================================================

lat = kwant.lattice.square(norbs=1)

# ============================================================
# BUILD SYSTEM
# ============================================================

sys = kwant.Builder()

# ------------------------------------------------------------
# Ring geometry
# ------------------------------------------------------------

for x in range(-R_outer, R_outer + 1):
    for y in range(-R_outer, R_outer + 1):

        r2 = x**2 + y**2

        if R_inner**2 < r2 < R_outer**2:
            sys[lat(x, y)] = 0

# ------------------------------------------------------------
# Hoppings
# ------------------------------------------------------------

sys[lat.neighbors()] = -1

# ============================================================
# LEFT LEAD
# ============================================================

left_lead = kwant.Builder(
    kwant.TranslationalSymmetry((-1, 0))
)

for y in range(-W_lead, W_lead + 1):
    left_lead[lat(0, y)] = 0

left_lead[lat.neighbors()] = -1

sys.attach_lead(left_lead)

# ============================================================
# RIGHT LEAD
# ============================================================

right_lead = kwant.Builder(
    kwant.TranslationalSymmetry((1, 0))
)

for y in range(-W_lead, W_lead + 1):
    right_lead[lat(0, y)] = 0

right_lead[lat.neighbors()] = -1

sys.attach_lead(right_lead.reversed())

# ============================================================
# FINALIZE
# ============================================================

fsys = sys.finalized()

# ============================================================
# CUSTOM COLORS (temporarily commented out due to compatibility issue)
# ============================================================

# def site_color(site):
#     x, y = site.pos

#     # Lead region
#     if abs(x) > R_outer - 2:
#         return 'lightcoral'

#     return 'black'

# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(8, 8))

kwant.plot(
    fsys,
    site_size=0.18,
    site_color='black', # Using a static color to resolve AttributeError
    hop_color='black'
    # bgcolor='white' # Removed this line
)

plt.show()
```

## Kwant Recommended Learning Order: Magnetic Field

```python
import kwant
import numpy as np
import matplotlib.pyplot as plt

# Define a square lattice
lat = kwant.lattice.square(norbs=1)

# Create a Kwant Builder
sys = kwant.Builder()

# Define the magnetic flux per unit cell (in units of flux quantum)
phi = 0.1

# Add sites in a 10x10 square
L = 10
for x in range(L):
    for y in range(L):
        sys[lat(x, y)] = 0 # On-site potential

# Define a hopping function with Peierls phase for a uniform magnetic field
def hopping_with_magnetic_field(site1, site2, phi):
    x1, y1 = site1.pos
    x2, y2 = site2.pos
    # Peierls phase for hopping from (x1, y1) to (x2, y2)
    # Here we use the Landau gauge A = (0, Bx)
    return -1 * np.exp(1j * 2 * np.pi * phi * (x1 * (y2 - y1) / 2 + x2 * (y1 - y2) / 2))

# Add hoppings with the magnetic field effect
sys[lat.neighbors()] = hopping_with_magnetic_field # Apply the hopping function

# Finalize the system
fsys = sys.finalized()

# Visualize the system. Note: Kwant's plot doesn't explicitly show phase, but the model has it.
plt.figure(figsize=(6, 6))
kwant.plot(
    fsys,
    site_size=0.2,
    site_color='black',
    hop_color='gray'
)
plt.title(f'Square Lattice with Magnetic Field ($\phi$ = {phi})') # Set title using plt.title
plt.show()
```

## Kwant Recommended Learning Order: Hofstadter Butterfly

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.colors import LinearSegmentedColormap
from scipy.linalg import eigvalsh
import warnings
warnings.filterwarnings("ignore")

# ─── Physical constants ───────────────────────────────────────────────────────
t   = 2.7        # nearest-neighbour hopping (eV)
a   = 0.246e-9   # graphene lattice constant (m)
e   = 1.6e-19    # electron charge (C)
hbar = 1.055e-34 # reduced Planck (J·s)
h   = 2 * np.pi * hbar

# ─── 1. HOFSTADTER BUTTERFLY  ─────────────────────────────────────────────────
def build_gnr_hamiltonian(N, phi, kx):
    """
    Build the tight-binding Hamiltonian for an N-site zigzag GNR unit cell
    at crystal momentum kx and flux quanta phi = Φ/Φ₀ per hexagon.

    Uses Peierls substitution:  t_ij → t · exp(i · 2π · phi · m)
    for hopping along y (between rows m).
    """
    H = np.zeros((2 * N, 2 * N), dtype=complex)

    for m in range(N):
        # Intra-dimer coupling (A–B within same row)
        peierls = np.exp(1j * 2 * np.pi * phi * m)
        H[2*m,   2*m+1] = -t * peierls
        H[2*m+1, 2*m  ] = -t * np.conj(peierls)

        # Inter-dimer coupling (B in row m → A in row m+1)
        if m < N - 1:
            peierls2 = np.exp(1j * 2 * np.pi * phi * (m + 0.5))
            H[2*m+1, 2*(m+1)  ] = -t * peierls2
            H[2*(m+1), 2*m+1  ] = -t * np.conj(peierls2)

    # kx-dependent hopping (periodic boundary along x)
    H[0,   2*N-1] += -t * np.exp(-1j * kx)
    H[2*N-1, 0  ] += -t * np.exp( 1j * kx)

    return H


def compute_butterfly(N=20, n_phi=200, n_kx=30):
    """Compute the Hofstadter butterfly spectrum for a ZGNR."""
    phi_vals = np.linspace(0, 1, n_phi)
    kx_vals  = np.linspace(-np.pi, np.pi, n_kx, endpoint=False)

    phi_plot = []
    E_plot   = []

    for phi in phi_vals:
        eigs_phi = []
        for kx in kx_vals:
            H = build_gnr_hamiltonian(N, phi, kx)
            eigs = eigvalsh(H)
            eigs_phi.extend(eigs)
        phi_plot.extend([phi] * len(eigs_phi))
        E_plot.extend(eigs_phi)

    return np.array(phi_plot), np.array(E_plot)


# ─── 2. LANDAU LEVELS (Dirac fermions) ───────────────────────────────────────
def landau_levels(phi_vals, n_max=5):
    """
    Relativistic Landau levels for massless Dirac fermions:
        E_n = sign(n) * v_F * sqrt(2 * e * hbar * B * |n|)
    In tight-binding units:
        E_n ≈ ±t * sqrt(6π |n| φ)
    """
    levels = {}
    for n in range(-n_max, n_max + 1):
        E_n = np.sign(n) * t * np.sqrt(6 * np.pi * np.abs(n) * phi_vals + 1e-12)
        levels[n] = E_n
    return levels


# ─── 3. EDGE STATES (zigzag flat band) ────────────────────────────────────────
def compute_edge_states(N=20, phi=0.1, n_kx=200):
    """
    Compute the band structure at fixed phi; identify edge-localised states
    by their probability weight on boundary sites.
    """
    kx_vals = np.linspace(-np.pi, np.pi, n_kx)
    bands   = np.zeros((n_kx, 2 * N))
    weights = np.zeros((n_kx, 2 * N))  # edge localisation weight

    for i, kx in enumerate(kx_vals):
        H = build_gnr_hamiltonian(N, phi, kx)
        evals, evecs = np.linalg.eigh(H)
        bands[i] = evals
        # Edge weight: probability on first + last 2 sites
        for j in range(2 * N):
            edge_sites = [0, 1, 2*N-2, 2*N-1]
            weights[i, j] = np.sum(np.abs(evecs[edge_sites, j])**2)

    return kx_vals, bands, weights


# ─── 4. CHERN NUMBERS (TKNN) ─────────────────────────────────────────────────
def chern_number_from_tknn(p, q):
    """
    Solve the Diophantine equation  p·s + q·t = 1  for s (Chern number)
    using the TKNN formula. Returns Chern number s for each of q gaps.
    """
    # Extended Euclidean algorithm
    def ext_gcd(a, b):
        if b == 0:
            return a, 1, 0
        g, x, y = ext_gcd(b, a % b)
        return g, y, x - (a // b) * y

    chern_numbers = []
    for gap_idx in range(1, q):
        # For gap r, solve: p·s ≡ r (mod q)
        g, s, _ = ext_gcd(p, q)
        s = (s * gap_idx) % q
        if s > q // 2:
            s -= q
        chern_numbers.append(s)
    return chern_numbers


# ─── 5. MAGNETOTRANSPORT ──────────────────────────────────────────────────────
def magnetotransport(phi_vals, T=1.0, delta=0.01):
    """
    Compute longitudinal (σxx) and Hall (σxy) conductances using
    a simplified Kubo formula model for a ZGNR.

    Hall conductance σxy = C · e²/h where C is integer (Chern number).
    σxx peaks at Landau level energies (broadened by disorder Γ=delta·t).
    """
    E_F = 0.0     # Fermi energy at charge neutrality point
    sigma_xy = []
    sigma_xx = []
    R_xx     = []
    R_xy     = []

    for phi in phi_vals:
        # Number of filled LLs below E_F
        # For Dirac LLs: E_n ≈ ±t*sqrt(6π|n|φ)
        n_filled = 0
        for n in range(1, 10):
            E_n = t * np.sqrt(6 * np.pi * n * phi)
            if E_n < -E_F:   # negative LL
                n_filled -= 1
            if -E_n < E_F:   # positive LL
                n_filled += 1

        # σxy = ν · e²/h, with ν = 4(n+1/2) for graphene (4-fold degeneracy)
        nu = 4 * n_filled
        sxy = nu   # in units of e²/h

        sxx = 0.0
        for n in range(-5, 6):
            E_n = np.sign(n) * t * np.sqrt(6 * np.pi * np.abs(n) * phi + 1e-6)
            sxx += delta / ((E_F - E_n)**2 + delta**2)
        sxx *= (e**2 / h)

        sigma_xy.append(sxy)
        sigma_xx.append(sxx)

        # Resistivity tensor inversion: ρ = σ⁻¹
        sxy_SI = sxy * (e**2 / h)
        denom  = sxy_SI**2 + sxx**2 + 1e-30
        R_xx.append(sxx / denom)
        R_xy.append(sxy_SI / denom)

    return (np.array(sigma_xx), np.array(sigma_xy),
            np.array(R_xx),     np.array(R_xy))


# ─── PLOTTING ─────────────────────────────────────────────────────────────────
def make_colormap():
    colors = ['#534AB7', '#EF9F27', '#0F6E56']   # purple → amber → teal
    return LinearSegmentedColormap.from_list('gnr', colors)


def plot_all():
    print("Computing Hofstadter butterfly  (N=20, 200 φ-values) …")
    phi_b, E_b = compute_butterfly(N=20, n_phi=200, n_kx=25)

    print("Computing edge states at φ=0.1 …")
    kx_e, bands_e, weights_e = compute_edge_states(N=20, phi=0.1)

    phi_vals = np.linspace(0.005, 1.0, 400)
    print("Computing Landau levels …")
    ll = landau_levels(phi_vals, n_max=4)

    print("Computing magnetotransport …")
    sxx, sxy, Rxx, Rxy = magnetotransport(phi_vals)

    cmap = make_colormap()

    fig = plt.figure(figsize=(16, 14), facecolor='#0d0d0b')
    fig.suptitle(
        "Magnetotransport & Topologically Protected Edge States\nin Graphene Nanoribbons — Hofstadter Butterfly Spectrum",
        color='white', fontsize=14, fontweight='bold', y=0.98
    )

    gs = gridspec.GridSpec(3, 2, figure=fig,
                           hspace=0.42, wspace=0.32,
                           left=0.08, right=0.96, top=0.93, bottom=0.06)

    ax1 = fig.add_subplot(gs[0, :])   # Butterfly (full width)
    ax2 = fig.add_subplot(gs[1, 0])   # Edge states band structure
    ax3 = fig.add_subplot(gs[1, 1])   # Landau levels
    ax4 = fig.add_subplot(gs[2, 0])   # Hall conductance
    ax5 = fig.add_subplot(gs[2, 1])   # Longitudinal resistance

    dark_bg = '#111110'
    for ax in [ax1, ax2, ax3, ax4, ax5]:
        ax.set_facecolor(dark_bg)
        ax.tick_params(colors='#aaa9a3', labelsize=8)
        for spine in ax.spines.values():
            spine.set_edgecolor('#333330')

    # ── Panel 1: Hofstadter Butterfly ────────────────────────────────────────
    emin, emax = E_b.min(), E_b.max()
    norm_e = (E_b - emin) / (emax - emin)
    colors_b = cmap(norm_e)

    ax1.scatter(phi_b, E_b, c=colors_b, s=0.3, alpha=0.6, rasterized=True)
    ax1.axhline(0, color='#ffffff44', lw=0.8, ls='--')
    ax1.set_xlabel('Magnetic flux per unit cell  φ = Φ/Φ₀', color='#ccc', fontsize=9)
    ax1.set_ylabel('Energy  E  (eV)', color='#ccc', fontsize=9)
    ax1.set_title('Hofstadter Butterfly — ZGNR (N = 20)', color='white', fontsize=10)
    ax1.set_xlim(0, 1)
    ax1.set_ylim(emin * 1.05, emax * 1.05)

    # Annotate rational flux values
    for p, q, label in [(1,3,'1/3'), (1,4,'1/4'), (1,2,'1/2'),
                         (2,3,'2/3'), (3,4,'3/4')]:
        ax1.axvline(p/q, color='#ffffff20', lw=0.6)
        ax1.text(p/q, emax * 0.88, label, color='#ffffff60',
                 ha='center', fontsize=7)

    # ── Panel 2: Edge State Band Structure ───────────────────────────────────
    for j in range(bands_e.shape[1]):
        w = weights_e[:, j]
        ax2.scatter(kx_e / np.pi, bands_e[:, j],
                    c=cmap(w), s=1.2, alpha=0.7, rasterized=True)

    ax2.axhline(0,  color='#ffffff44', lw=0.8, ls='--')
    ax2.axvline(-2/3, color='#ffffff30', lw=0.6)
    ax2.axvline( 2/3, color='#ffffff30', lw=0.6)
    ax2.text(-2/3, bands_e.max()*0.8, 'K', color='#aaa', ha='center', fontsize=7)
    ax2.text( 2/3, bands_e.max()*0.8, "K'", color='#aaa', ha='center', fontsize=7)
    ax2.set_xlabel('Crystal momentum  kₓ / π', color='#ccc', fontsize=9)
    ax2.set_ylabel('Energy  E  (eV)', color='#ccc', fontsize=9)
    ax2.set_title('Band Structure at φ = 0.1\n(colour = edge localisation weight)',
                  color='white', fontsize=9)
    ax2.set_xlim(-1, 1)

    # Colorbar for edge weight
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(0, 1))
    cb = plt.colorbar(sm, ax=ax2, pad=0.02)
    cb.set_label('Edge weight', color='#aaa', fontsize=8)
    cb.ax.yaxis.set_tick_params(color='#aaa', labelsize=7)
    plt.setp(cb.ax.yaxis.get_ticklabels(), color='#aaa')

    # ── Panel 3: Landau Levels ────────────────────────────────────────────────
    ll_colors = {
        0:  '#EF9F27',
        1:  '#5DCAA5',  -1: '#5DCAA5',
        2:  '#7F77DD',  -2: '#7F77DD',
        3:  '#D85A30',  -3: '#D85A30',
        4:  '#378ADD',  -4: '#378ADD',
    }
    for n, E_n in ll.items():
        mask = np.abs(E_n) < 3 * t
        col  = ll_colors.get(n, '#888')
        lbl  = f'n = {n}' if n >= 0 else None
        ax3.plot(phi_vals[mask], E_n[mask], color=col, lw=1.3,
                 alpha=0.9, label=lbl)

    ax3.axhline(0, color='#ffffff55', lw=1.2, ls='--', label='n = 0  (E=0)')
    ax3.set_xlabel('φ = Φ/Φ₀', color='#ccc', fontsize=9)
    ax3.set_ylabel('Landau level energy  E  (eV)', color='#ccc', fontsize=9)
    ax3.set_title('Relativistic Landau Levels\nEₙ = ±v_F √(2eℏB|n|)', color='white', fontsize=9)
    ax3.set_xlim(0, 1)
    ax3.set_ylim(-3*t, 3*t)
    leg = ax3.legend(fontsize=7, loc='upper left', framealpha=0.2,
                     labelcolor='white', facecolor='#111')
    leg.get_frame().set_edgecolor('#333')

    # ── Panel 4: Hall conductance σxy / (e²/h) ───────────────────────────────
    ax4.plot(phi_vals, sxy, color='#7F77DD', lw=1.4, alpha=0.9)
    ax4.axhline(0,  color='#ffffff22', lw=0.6)
    for nu in [-6, -2, 2, 6, 10]:
        ax4.axhline(nu, color='#ffffff15', lw=0.5, ls=':')
        ax4.text(0.98, nu, f'ν={nu}', color='#ffffff44',
                 ha='right', va='center', fontsize=6.5)
    ax4.set_xlabel('φ = Φ/Φ₀', color='#ccc', fontsize=9)
    ax4.set_ylabel('σ_xy  (e²/h)', color='#ccc', fontsize=9)
    ax4.set_title('Quantum Hall Conductance\n(integer plateaus — TKNN Chern #)',
                  color='white', fontsize=9)
    ax4.set_xlim(0, 1)

    # ── Panel 5: Longitudinal resistance Rxx ─────────────────────────────────
    ax5.plot(phi_vals, Rxx * 1e3, color='#F0997B', lw=1.2, alpha=0.9,
             label='R_xx (mΩ)')
    ax5_r = ax5.twinx()
    ax5_r.plot(phi_vals, np.abs(Rxy), color='#5DCAA5', lw=1.1,
               alpha=0.7, ls='--', label='|R_xy| (Ω)')
    ax5_r.tick_params(colors='#aaa9a3', labelsize=8)
    ax5_r.set_ylabel('|R_xy|  (Ω)', color='#5DCAA5', fontsize=8)
    for spine in ax5_r.spines.values():
        spine.set_edgecolor('#333330')

    ax5.set_xlabel('φ = Φ/Φ₀', color='#ccc', fontsize=9)
    ax5.set_ylabel('R_xx  (mΩ)', color='#F0997B', fontsize=9)
    ax5.set_title('Magnetotransport: Shubnikov–de Haas\n(R_xx oscillations, Hall R_xy)',
                  color='white', fontsize=9)
    ax5.set_xlim(0, 1)

    lines1, labs1 = ax5.get_legend_handles_labels()
    lines2, labs2 = ax5_r.get_legend_handles_labels()
    leg2 = ax5.legend(lines1 + lines2, labs1 + labs2, fontsize=7,
                      loc='upper right', framealpha=0.2,
                      labelcolor='white', facecolor='#111')
    leg2.get_frame().set_edgecolor('#333')

    # ── TKNN annotation ───────────────────────────────────────────────────────
    fig.text(0.5, 0.005,
             "TKNN theorem: σ_xy = Σ_n C_n · e²/h   |   "
             "Chern #  C_n from Diophantine eq.  p·s + q·t = 1   |   "
             "Edge states protected by bulk topology",
             ha='center', color='#555550', fontsize=7.5)

    plt.savefig('hofstadter_gnr.png', # Changed path to local file
                dpi=180, bbox_inches='tight', facecolor=fig.get_facecolor())
    print("Saved → hofstadter_gnr.png")
    plt.close()


# ─── CHERN NUMBER TABLE ───────────────────────────────────────────────────────
def print_chern_table():
    print("\n── TKNN Chern numbers for rational flux φ = p/q ──")
    print(f"{'p/q':>8}  {'Gap index':>10}  {'Chern #':>9}")
    print("─" * 34)
    for p, q in [(1,3),(1,4),(1,5),(2,5),(1,2),(3,5),(2,3),(3,4)]:
        from math import gcd
        if gcd(p, q) != 1:
            continue
        cherns = chern_number_from_tknn(p, q)
        for idx, C in enumerate(cherns, 1):
            print(f"  {p}/{q}     gap {idx:>3}          C = {C:>+3}")
    print()


# ─── MAIN ─────────────────────────────────────────────────────────────────────
if __name__ == '__main__':
    print_chern_table()
    plot_all()
    print("Done.")
```

## Kwant Recommended Learning Order: Topological Systems

Topological systems are a fascinating area of condensed matter physics where the bulk properties of a material are trivial (e.g., insulating), but its boundaries or defects host robust, non-trivial states. These states are 'topologically protected,' meaning they are immune to small perturbations like disorder or defects.

In the context of Kwant, one can study topological systems by constructing tight-binding models of materials exhibiting such phases. Examples include:

*   **Quantum Hall Effect (QHE)**: Characterized by quantized Hall conductance and chiral edge states. The Hofstadter butterfly, as shown above, is a manifestation of the QHE in a lattice.
*   **Quantum Spin Hall Effect (QSHE)**: An insulating bulk with spin-filtered counter-propagating edge states, protected by time-reversal symmetry.
*   **Topological Insulators**: Materials that are electrical insulators in their interior but have conducting states on their surface or edges.
*   **Weyl and Dirac Semimetals**: Materials with protected band crossings in their bulk.

Kwant's ability to build complex systems and calculate transport properties (like scattering matrices and conductance) makes it an excellent tool for investigating these systems. One typically looks for gapped bulk spectra and analyzes edge modes or quantized transport to identify topological phases.

## Kwant Recommended Learning Order: Spin-Orbit Coupling

Spin-orbit coupling (SOC) is an interaction between an electron's spin and its orbital motion. In condensed matter systems, SOC can significantly alter band structures and is a key ingredient in many topological phenomena, particularly in topological insulators and the quantum spin Hall effect.

In Kwant, SOC terms are implemented as additional hopping terms in the tight-binding Hamiltonian. These hoppings are typically spin-dependent and often depend on the relative positions of the sites involved. There are several types of SOC:

*   **Rashba SOC**: Often arises at surfaces or interfaces due to structural inversion asymmetry. It leads to spin splitting of energy bands.
*   **Dresselhaus SOC**: Arises from bulk inversion asymmetry in certain crystal structures.
*   **Intrinsic SOC (Kane-Mele type)**: Present in materials like graphene, it opens a gap at the Dirac points and leads to a quantum spin Hall phase.

Implementing SOC in Kwant involves:

1.  **Defining a lattice with spin degrees of freedom**: This means `norbs` in `kwant.lattice` would be at least 2 (for spin-up and spin-down).
2.  **Modifying hopping functions**: The `hopping` functions (or `sys[lat.neighbors()]`) would become matrices in spin space (e.g., 2x2 matrices for spin-1/2 particles) that include `sigma_x`, `sigma_y`, `sigma_z` Pauli matrices to represent the spin interaction terms.
3.  **Analyzing band structure and edge states**: Look for spin-polarized bands or edge states characteristic of SOC-driven topological phases.

## Kwant Recommended Learning Order: Topological Systems - SSH Model

The Su-Schrieffer-Heeger (SSH) model is a one-dimensional tight-binding model that serves as a paradigmatic example of a topological insulator. It describes a chain of atoms with alternating hopping strengths, which can exhibit different topological phases depending on these strengths. In one phase, the chain is trivial, while in the other, it hosts robust edge states.

Here, we will build an SSH chain and visualize its structure. While Kwant itself does not directly compute topological invariants, it is an excellent tool for constructing such systems and then using its scattering matrix functionalities to probe their topological properties (e.g., observing quantized conductance due to edge states in a transport setup).

```python
import kwant
import matplotlib.pyplot as plt
import numpy as np

# Define the hopping strengths
t_1 = 1.0  # Intracell hopping
t_2 = 0.5  # Intercell hopping

# Number of unit cells
N = 10

# Define a 1D lattice
sym = kwant.lattice.chain(norbs=1)

# Create a Kwant Builder
sys = kwant.Builder()

# 1. Add all sites first
for i in range(N):
    # Add sites to the unit cell (A and B sublattices)
    sys[sym(2 * i)] = 0  # Site A
    sys[sym(2 * i + 1)] = 0  # Site B

# 2. Add all hoppings after all sites are added
for i in range(N):
    # Intracell hopping
    sys[sym(2 * i), sym(2 * i + 1)] = -t_1

    # Intercell hopping (if not the last unit cell)
    if i < N - 1:
        sys[sym(2 * i + 1), sym(2 * i + 2)] = -t_2

# Finalize the system
fsys = sys.finalized()

# Plot the system
plt.figure(figsize=(10, 3))
kwant.plot(
    fsys,
    site_size=0.3,
    site_color='blue',
    hop_color='red',
    pos_transform=lambda p: p + (0.5 * (-1)**(p[0] % 2), 0) # Slightly offset A/B sites for visual clarity
)
plt.title(f'1D SSH Chain (t1={t_1}, t2={t_2})')
plt.xlabel('Site Index')
plt.ylabel('')
plt.yticks([])
plt.show()
```

## Kwant Recommended Learning Order: Spin-Orbit Coupling - Rashba SOC

Spin-orbit coupling (SOC) describes the interaction between an electron's spin and its motion. It's a crucial ingredient in many topological phenomena and spintronic applications. In Kwant, implementing SOC typically involves defining lattices with multiple orbitals (representing spin degrees of freedom) and then modifying hopping terms to be spin-dependent matrices, often using Pauli matrices.

Here, we will demonstrate how to introduce Rashba spin-orbit coupling into a 2D square lattice. Rashba SOC typically arises in systems lacking inversion symmetry and couples the electron's momentum to its spin, leading to spin splittings in the band structure. This example will show how to construct such a Hamiltonian within Kwant.

```python
import kwant
import numpy as np
import matplotlib.pyplot as plt
from scipy.sparse import csc_matrix

# Pauli matrices for spin-1/2
sx = csc_matrix([[0, 1], [1, 0]])
sy = csc_matrix([[0, -1j], [1j, 0]])
sz = csc_matrix([[1, 0], [0, -1]])
s0 = csc_matrix([[1, 0], [0, 1]])

# Define the lattice with 2 orbitals (spin up and down) per site
lat = kwant.lattice.square(norbs=2)  # Two orbitals for spin

# Create a Kwant Builder
sys = kwant.Builder()

# System dimensions
W = 5  # Width
L = 5  # Length

# On-site potential (diagonal in spin space)
for x in range(L):
    for y in range(W):
        sys[lat(x, y)] = 0.5 * sz  # Example: add a Zeeman-like term

# Hopping function for Rashba SOC
def rashba_hopping(site1, site2, t, alpha):
    # Basic hopping
    hop = -t * s0

    # Vector connecting sites
    dx, dy = site1.pos - site2.pos

    # Rashba SOC term: alpha * (sigma_x * dy - sigma_y * dx)
    # Note: Kwant expects hoppings from 'from_site' to 'to_site'.
    # The vector (dx, dy) is from site2 to site1.
    # So for hopping from site2 to site1, the momentum component is (dx, dy).
    # The conventional Rashba term is proportional to (k_y * sigma_x - k_x * sigma_y).
    # If dx is along x and dy along y, then k_x ~ dx and k_y ~ dy.
    # Thus, hopping in x-direction (dx!=0, dy=0) needs sigma_y component
    # And hopping in y-direction (dx=0, dy!=0) needs sigma_x component

    if dx == -1: # Hopping from (x+1, y) to (x, y) (i.e. site1 is to the left of site2)
        hop += 1j * alpha * sy
    elif dx == 1: # Hopping from (x-1, y) to (x, y) (i.e. site1 is to the right of site2)
        hop -= 1j * alpha * sy
    elif dy == -1: # Hopping from (x, y+1) to (x, y) (i.e. site1 is below site2)
        hop -= 1j * alpha * sx
    elif dy == 1: # Hopping from (x, y-1) to (x, y) (i.e. site1 is above site2)
        hop += 1j * alpha * sx

    return hop

# Parameters for hoppings
t_hop = 1.0   # Normal hopping strength
alpha_R = 0.3 # Rashba SOC strength

# Add hoppings to the system (use a dictionary to pass parameters)
sys[lat.neighbors()] = lambda site1, site2: rashba_hopping(site1, site2, t=t_hop, alpha=alpha_R)

# Finalize the system
fsys = sys.finalized()

# Kwant's default plot does not visualize spin degrees of freedom directly.
# To visualize, one would typically compute properties like local density of states
# resolved by spin, or plot the spin texture in momentum space.
# For now, we'll just plot the basic connectivity.

plt.figure(figsize=(6, 6))
kwant.plot(
    fsys,
    site_size=0.25,
    site_color='grey', # Color does not reflect spin here
    hop_color='black'
)
plt.title(f'Square Lattice with Rashba SOC (alpha={alpha_R})')
plt.show()

print("System with Rashba SOC built. Note: Kwant's plot does not show spin degrees of freedom directly.")
```

### Conductance of a 1D Quantum Wire
This example demonstrates how to:
1. Define a 1D lattice.
2. Create a scattering region (the wire).
3. Attach infinite leads.
4. Calculate and plot the conductance as a function of energy.

```python
import kwant
import matplotlib.pyplot as plt
import numpy as np

def make_system(a=1, t=1.0, W=10, L=30):
    # 1. Define the lattice (square lattice with unit spacing)
    lat = kwant.lattice.square(a, norbs=1)
    sys = kwant.Builder()

    # 2. Define the scattering region
    def box_shape(pos):
        x, y = pos
        return 0 <= x < L and 0 <= y < W

    sys[lat.shape(box_shape, (0, 0))] = 4 * t
    sys[lat.neighbors()] = -t

    # 3. Define and attach leads
    lead = kwant.Builder(kwant.TranslationalSymmetry((-a, 0)))
    def lead_shape(pos):
        x, y = pos
        return 0 <= y < W

    lead[lat.shape(lead_shape, (0, 0))] = 4 * t
    lead[lat.neighbors()] = -t

    sys.attach_lead(lead)
    sys.attach_lead(lead.reversed())

    return sys

sys = make_system()
kwant.plot(sys)
fsys = sys.finalized()

# 4. Calculate Conductance
energies = np.linspace(-0.5, 3.5, 100)
data = []
for energy in energies:
    smatrix = kwant.smatrix(fsys, energy)
    data.append(smatrix.transmission(1, 0))

plt.figure()
plt.plot(energies, data)
plt.xlabel("Energy [t]")
plt.ylabel("Conductance [e^2/h]")
plt.title("1D Wire Conductance (Quantized Steps)")
plt.show()
```
